**Find duplicate number in interger list**:

```python
def find_duplicates(elements: list[int]) -> set[int]:
    seen = set()
    duplicates = set()
    for element in elements:
        if element in seen:
            duplicates.add(element)
        else:
            seen.add(element)
    return duplicates
```

**Check if two strings are anagrams**:

```python
def is_anagram(s1: str, s2: str) -> bool:
    return sorted(s1) == sorted(s2)

print("['listen', 'silent'] is anagram:", is_anagram("listen", "silent"))  # True
print("['hello', 'world'] is anagram:", is_anagram("hello", "world"))    # False

is_anagram_lambda = lambda s1, s2: sorted(s1) == sorted(s2)

print("['angered', 'enraged'] is anagram:", is_anagram_lambda("listen", "silent"))
print("['hello', 'world'] is anagram:", is_anagram_lambda("hello", "world"))
```

**Check if two strings are anagrams without sorting** (Standard Interview Logic):
* **Early Exit**: Anagrams must be identical in length.
* **Net-Zero Frequency**: Characters in anagrams must balance out to zero.
* **Safe Map Retrieval**: Using count.get(char, 0)


```python

def is_anagram_count(s1: str, s2: str) -> bool:
    if len(s1) != len(s2):
        return False
    count = {}
    for char in s1:
        count[char] = count.get(char, 0) + 1
    for char in s2:
        if char not in count or count[char] == 0:
            return False
        count[char] -= 1
    return True

print("['elbow', 'below'] is anagram:", is_anagram_count("elbow", "below"))
```

Merging into One Unique List
* Using set union operation

```python
def merge_unique(lst1: list[int], lst2: list[int]) -> list[int]:
    return list(set(lst1) | set(lst2))
```

* Using dictionary hashing

```python
def merge_unique_dict(lst1: list[int], lst2: list[int]) -> list[int]:
    unique_dict = {}
    for item in lst1 + lst2:
        unique_dict[item] = True
    return list(unique_dict.keys())
```

Subtracting Duplicates (Difference)
* To remove duplicates from list1 that also exist in list2"

```python
def subtract_duplicates(lst1: list[int], lst2: list[int]) -> list[int]:
    set2 = set(lst2)
    return [item for item in lst1 if item not in set2]
```

**Check if a string is a palindrome**:

```python
def is_palindrome(s: str) -> bool:
    return s == s[::-1]

# Alternative method using two-pointer technique for memory efficiency
def is_palindrome_two_pointer(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True
```

**Get missing number in list of integers from [1...100]**:

```python
def find_missing_number(lst: list[int], n: int) -> int:
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(lst)
    return expected_sum - actual_sum

def get_missing_numbers(lst: list[int], n: int) -> list[int]:
    full_set = set(range(1, n + 1))
    current_set = set(lst)
    missing_numbers = list(full_set - current_set)
    return missing_numbers

lst = list(range(1, 101))
lst.remove(55)
print("Missing number is:", find_missing_number(lst, 100))

lst = list(range(1, 101))
lst.remove(20)
lst.remove(75)
print("Missing numbers are:", get_missing_numbers(lst, 100))
```

**Compute the intersection of two lists**:

```python
def intersection(lst1: list[int], lst2: list[int]) -> list[int]:
    seen, work_list = [], lst2[:]
    for item in lst1:
        if item in work_list:
            seen.append(item)
            work_list.remove(item)  # To avoid counting duplicates
    return seen
print("Intersection is:", intersection([1,2,3,4,5], [4,5,6,7,8]))
```

**Reverse a string using recursion**:

```python
def reverse_string(s: str) -> str:
    if len(s) == 0:
        return s
    else:
        return s[-1] + reverse_string(s[:-1])

print("Reversed string is:", reverse_string("hello world"))
```

**Sort list with QuickSort algorithm**:

```python
def quicksort(arr: list[int]) -> list[int]:
    if not arr:
        return []
    else:
        pivot = arr[0]
        less_than_pivot = [x for x in arr[1:] if x <= pivot]
        greater_than_pivot = [x for x in arr[1:] if x > pivot]
        return quicksort(less_than_pivot) + [pivot] + quicksort(greater_than_pivot)

lst = [4,3,5,1,2,8,-7,22, 3]
sorted_lst = quicksort(lst)
print("Sorted list is:", sorted_lst)
```

**Find all permuatations of a string using recursion**:

```python
def get_permutations(s: str) -> list[str]:
    if len(s) == 0:
        return ['']
    smaller = get_permutations(s[1:])
    permutations = set()
    for perm in smaller:
        for i in range(len(perm) + 1):
            new_perm = perm[:i] + s[0] + perm[i:]
            permutations.add(new_perm)
    return list(permutations)

print("Permutations are:", get_permutations("abc"))
print("Number of permutations:", get_permutations("nan"))
```

**Create a class with metaclass that increments a class variable x by 1 for each subclass created**:
- `pass` statement is used to indicate that the class does not add any new attributes or methods beyond those inherited from its parent class.

```python
class Meta(type):
    def __new__(cls, name, bases, dct):
        print(f"Creating class {name} with Meta")
        dct["x"] = dct.get("x", 0) + 1
        return super().__new__(cls, name, bases, dct)

class A(metaclass=Meta):
    x = 5

class B(A):
    pass

print("A.x =", A.x, ", B.x =", B.x)
```

**function mutable default argument pitfall**:

```python
# Problematic function with mutable default argument
def func_problematic(k: str, v: str, my_dict = {}) -> dict[str, str]:
    if my_dict is None:
        my_dict = {}
    my_dict[k] = v
    return my_dict

print(func_problematic("a", "1"))  # {'a': '1'}
print(func_problematic("b", "2"))  # {'a': '1', 'b': '2'}
```

```python
def func_safe(k: str, v: str, my_dict: dict[str, str] | None = None) -> dict[str, str]:
    if my_dict is None:
        my_dict = {}
    my_dict[k] = v
    return my_dict

print(func_safe("a", "1"))  # {'a': '1'}
print(func_safe("b", "2"))  # {'b': '2'}
```

**class __get__ method tricky example Vanish **:

**Key insight**: The __get__ descriptor method is destructive—it deletes the attribute it's supposed to manage. The first access succeeds and removes itself, making the second access fail.

Here's the step-by-step breakdown:
1. First `print(a.x, hasattr(a, 'x'))`:
   - Accessing `a.x` triggers the `__get__` method of the Vanish descriptor
   - Inside `__get__`, the line `del owner.x` deletes the class variable x from class A
   - The method returns 100, so this prints `100 False` (because x no longer exists after deletion)
2. Second `print(a.x, hasattr(a, 'x'))`:
   - Python tries to access `a.x` again
   - Since `x` was deleted from class `A`, Python searches for it but can't find it
   - This raises an `AttributeError: type object 'A' has no attribute 'x'`
   - The exception is caught by your `try-except` block, printing the error message

```python
class Vanish:
    def __get__(self, obj, owner):
        del owner.x
        return 100
class A:
    x = Vanish()

a = A()
# print("a.x =", a.x)  # 100
print(a.x, hasattr(a, 'x'))  # False
try:
    print(a.x, hasattr(a, 'x'))
except AttributeError as e:
    print("Error", e)
```

**class __init__**:
 - __init__ method initializes instance variables, not class variables.

```python
class A:
    count = 0
    def __init__(self):
        print("A init called")
        self.count +=1

a = A()
b = A()
print("a.count = ", a.count, ", b.count =", b.count, ", A.count =", A.count)  # 1 1 0
```
