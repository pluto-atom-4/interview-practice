**Find duplicate number in interger list**:

```python
def find_duplicates(elements: list[int]) -> set[int]:
    seen = set()
    duplicates = set()
    for element in elements:
        if element in seen:
            duplicates.add(element)
        else:
            seen.add(element)
    return duplicates
```

**Check if two strings are anagrams**:

```python
def is_anagram(s1: str, s2: str) -> bool:
    return sorted(s1) == sorted(s2)

print("['listen', 'silent'] is anagram:", is_anagram("listen", "silent"))  # True
print("['hello', 'world'] is anagram:", is_anagram("hello", "world"))    # False

is_anagram_lambda = lambda s1, s2: sorted(s1) == sorted(s2)

print("['angered', 'enraged'] is anagram:", is_anagram_lambda("listen", "silent"))
print("['hello', 'world'] is anagram:", is_anagram_lambda("hello", "world"))
```

**Check if a string is a palindrome**:

```python
def is_palindrome(s: str) -> bool:
    return s == s[::-1]

# Alternative method using two-pointer technique for memory efficiency
def is_palindrome_two_pointer(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True
```

**Use list as stack, array, and queue**:

```python
# Using a list as a stack
stack = [3, 4, 5]
print("Initial stack:", stack)
stack += [6]  # Push
print("Stack after push:", stack)

# stack pop
lst = [3, 4, 5, 6]
print("Stack:", lst)
lst.append(10)
print("List after pushing:", lst)
lst.pop()
print("List after popping:", lst)

# Using a list as a queue
lst.insert(0, 5)
print("Queue after pushing:", lst)
lst.pop(0)
print("Queue after polling:", lst)
```

**Get missing number in list of integers from [1...100]**:

```python
def find_missing_number(lst: list[int], n: int) -> int:
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(lst)
    return expected_sum - actual_sum

def get_missing_numbers(lst: list[int], n: int) -> list[int]:
    full_set = set(range(1, n + 1))
    current_set = set(lst)
    missing_numbers = list(full_set - current_set)
    return missing_numbers

lst = list(range(1, 101))
lst.remove(55)
print("Missing number is:", find_missing_number(lst, 100))

lst = list(range(1, 101))
lst.remove(20)
lst.remove(75)
print("Missing numbers are:", get_missing_numbers(lst, 100))
```

**Compute the intersection of two lists**:

```python
def intersection(lst1: list[int], lst2: list[int]) -> list[int]:
    seen, work_list = [], lst2[:]
    for item in lst1:
        if item in work_list:
            seen.append(item)
            work_list.remove(item)  # To avoid counting duplicates
    return seen
print("Intersection is:", intersection([1,2,3,4,5], [4,5,6,7,8]))
```

**Reverse a string using recursion**:

```python
def reverse_string(s: str) -> str:
    if len(s) == 0:
        return s
    else:
        return s[-1] + reverse_string(s[:-1])

print("Reversed string is:", reverse_string("hello world"))
```

**Sort list with QuickSort algorithm**:

```python
def quicksort(arr: list[int]) -> list[int]:
    if not arr:
        return []
    else:
        pivot = arr[0]
        less_than_pivot = [x for x in arr[1:] if x <= pivot]
        greater_than_pivot = [x for x in arr[1:] if x > pivot]
        return quicksort(less_than_pivot) + [pivot] + quicksort(greater_than_pivot)

lst = [4,3,5,1,2,8,-7,22, 3]
sorted_lst = quicksort(lst)
print("Sorted list is:", sorted_lst)
```

**Find all permuatations of a string using recursion**:

```python
def get_permutations(s: str) -> list[str]:
    if len(s) == 0:
        return ['']
    smaller = get_permutations(s[1:])
    permutations = set()
    for perm in smaller:
        for i in range(len(perm) + 1):
            new_perm = perm[:i] + s[0] + perm[i:]
            permutations.add(new_perm)
    return list(permutations)

print("Permutations are:", get_permutations("abc"))
print("Number of permutations:", get_permutations("nan"))
```

**Flatten a list of lists into a single list**:

```python
from functools import reduce

lst = [[3], [4, 5], [7]]

flatten = reduce(lambda x, y: x + y, lst)

# Output result
print(f"Flattened List: {flatten}")
```
