```python
from skyfield.api import load

planets = load('de421.bsp')
earth, mars = planets['earth'], planets['mars']

ts = load.timescale()
t = ts.now()
position = earth.at(t).observe(mars)
ra, dec, distance = position.radec()

print(ra)
print(dec)
print(distance)
```

```python
from skyfield.api import load
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

sat = load.tle_file('http://celestrak.com/NORAD/elements/stations.txt')[0]
ts = load.timescale()
times = ts.utc(2026, 1, 1, range(0,90,5))

lats, lons = [], []
for t in times:
    geo = sat.at(t).subpoint()
    lats.append(geo.latitude.degrees)
    lons.append(geo.longitude.degrees)

ax = plt.axes(projection=ccrs.PlateCarree())
ax.stock_img()
ax.plot(lons, lats,color='red')
plt.title("ISS Ground Track on Jan 1, 2026")
plt.show()
    
```
