```python
from skyfield.api import load

planets = load('de421.bsp')
earth, mars = planets['earth'], planets['mars']

ts = load.timescale()
t = ts.now()
position = earth.at(t).observe(mars)
ra, dec, distance = position.radec()

print(ra)
print(dec)
print(distance)
```

```python
from skyfield.api import load
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

sat = load.tle_file('http://celestrak.com/NORAD/elements/stations.txt')[0]
ts = load.timescale()
times = ts.utc(2026, 1, 1, range(0,90,5))

lats, lons = [], []
for t in times:
    geo = sat.at(t).subpoint()
    lats.append(geo.latitude.degrees)
    lons.append(geo.longitude.degrees)

ax = plt.axes(projection=ccrs.PlateCarree())
ax.stock_img()
ax.plot(lons, lats,color='red')
plt.title("ISS Ground Track on Jan 1, 2026")
plt.show()
```

**Electic Curcuit Diagram**:

```python
import schemdraw
import schemdraw.elements as elm

with schemdraw.Drawing(
    unit=2,
    fontsize=12,
    color='white',
    bg='black',
    ) as d:
    d += (V := elm.SourceV().up().label('10V'))
    d += elm.Resistor().right().label('R1\n100Ω')
    d += elm.Capacitor().down().label('C1\n10μF')
    d += elm.Resistor().left().label('R2\n200Ω')
    d += elm.Ground()
    d += elm.Line().up().to(V.start)

    d.draw()
```

```python
from PIL import Image, ImageDraw, ImageFont
from IPython.display import display

i = Image.new('RGB', (400, 200), color = (135, 206, 235))
d = ImageDraw.Draw(i)

def k(x, y, s, c):
    h = s//2
    d.polygon(([x, y - h], [x + h, y], [x, y + h], [x - h, y]), fill=c)

k(50, 50, 100, (255, 0, 0))
k(200, 50, 100, (0, 255, 0))

t = "Happy New Year of Horse!"

try:
    f = ImageFont.truetype("arial.ttf", 24)
except:
    f = ImageFont.load_default()

w, h = d.multiline_textbbox((0, 0), t, font=f)[2:]

d.multiline_text(((400 - w)//2, 150), t, fill=(255, 255, 255), font=f, align='center')
display(i)
```

**Heatmap of Brightness**:

```python
import cv2
import numpy as np
from urllib.request import urlopen

url = "https://wallpaperaccess.com/full/124129.jpg"
resp = urlopen(url)
img_array = np.asarray(bytearray(resp.read()), dtype=np.uint8)
img = cv2.imdecode(img_array, cv2.IMREAD_GRAYSCALE)

heatmap = cv2.applyColorMap(img, cv2.COLORMAP_JET)
cv2.imshow('Heatmap', heatmap)
cv2.waitKey(0)
```
